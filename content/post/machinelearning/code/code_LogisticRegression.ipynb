```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import log_loss
```

```python
np.random.seed(42)
n = 100
X = np.linspace(0, 10, n).reshape(-1, 1)
y = (X[:,0]> 5).astype(int)
```

```python
model = LogisticRegression()
```

```python
model.fit(X, y)
```

```python
y_pred_prob = model.predict_proba(X)[:,1]
```

```python
bce = log_loss(y, y_pred_prob)
print(f"BCE损失 = {bce:.4f}")
```

```python
x_curve = np.linspace(0,10,n).reshape(-1,1)
p = model.predict_proba(x_curve)[:,1]
plt.scatter(X, y, c="red")
plt.scatter(x_curve, p)
plt.xlabel("x")
plt.ylabel("P(y=1|x)")
plt.show()
```
