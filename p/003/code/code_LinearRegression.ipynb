```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
np.random.seed(42)
# sklearn 的特征 X 要求是二维：[样本数，特征数]
X = np.linspace(0, 5, 100).reshape(-1, 1)
y = 2 * X + 3 + np.random.normal(0, 0.5, (X.shape[0],1))
```

```python
plt.scatter(X, y, label="带噪声样本")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
model = LinearRegression()
model.fit(X, y)
```

```python
y_pred = model.predict(X)
```

```python
mse = mean_squared_error(y, y_pred)
print(f"MSE = {mse:.4f}")
```

```python
plt.scatter(X, y, label="带噪声样本")
plt.xlabel("x")
plt.ylabel("y")
plt.plot(X, y_pred, color="red", lw=2, label="拟合直线")
plt.legend()
plt.show()
```
